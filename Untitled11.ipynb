```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("Libraries imported successfully!")
```

```python
np.random.seed(42)

n = 500

study_hours = np.random.uniform(0, 10, n)
attendance = np.random.uniform(50, 100, n)
previous_score = np.random.uniform(30, 100, n)
assignments_completed = np.random.randint(0, 11, n)

noise = np.random.normal(0, 5, n)

exam_score = (
    3 * study_hours
    + 0.20 * attendance
    + 0.35 * previous_score
    + 1.2 * assignments_completed
    + noise
)

exam_score = np.clip(exam_score, 0, 100)

df = pd.DataFrame({
    "study_hours": study_hours.round(1),
    "attendance": attendance.round(1),
    "previous_score": previous_score.round(1),
    "assignments_completed": assignments_completed,
    "exam_score": exam_score.round(1)
})

print("Dataset created successfully!")
display(df.head(10))
```

```python
print("Dataset shape:", df.shape)

print("\nColumn information:")
df.info()

print("\nStatistical summary:")
display(df.describe())

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    df["exam_score"],
    bins=20,
    kde=True
)

plt.title("Distribution of Student Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="study_hours",
    y="exam_score"
)

plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours Per Day")
plt.ylabel("Exam Score")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Student Features")
plt.show()
```

```python
X = df[
    [
        "study_hours",
        "attendance",
        "previous_score",
        "assignments_completed"
    ]
]

y = df["exam_score"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
linear_model = LinearRegression()

linear_model.fit(X_train, y_train)

print("Linear Regression trained successfully!")

random_forest_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

random_forest_model.fit(X_train, y_train)

print("Random Forest trained successfully!")
```

```python
models = {
    "Linear Regression": linear_model,
    "Random Forest": random_forest_model
}

results = []

for name, model in models.items():
    predictions = model.predict(X_test)

    mae = mean_absolute_error(y_test, predictions)
    rmse = np.sqrt(mean_squared_error(y_test, predictions))
    r2 = r2_score(y_test, predictions)

    results.append({
        "Model": name,
        "MAE": round(mae, 2),
        "RMSE": round(rmse, 2),
        "R2 Score": round(r2, 3)
    })

results_df = pd.DataFrame(results)

display(results_df)
```

```python
results_df.set_index("Model")[["MAE", "RMSE"]].plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Comparison of Prediction Errors")
plt.ylabel("Error in Marks")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
best_model_name = (
    results_df.sort_values("RMSE").iloc[0]["Model"]
)

best_model = models[best_model_name]

y_pred = best_model.predict(X_test)

print("Selected model:", best_model_name)

plt.figure(figsize=(7, 6))

plt.scatter(y_test, y_pred, alpha=0.7)

plt.plot([0, 100], [0, 100], linestyle="--")

plt.xlabel("Actual Exam Score")
plt.ylabel("Predicted Exam Score")
plt.title("Actual vs Predicted Student Scores")
plt.xlim(0, 100)
plt.ylim(0, 100)
plt.show()
```

```python
importance = pd.Series(
    random_forest_model.feature_importances_,
    index=X.columns
).sort_values()

importance.plot(
    kind="barh",
    figsize=(8, 5)
)

plt.title("Feature Importance for Student Score Prediction")
plt.xlabel("Relative Importance")
plt.tight_layout()
plt.show()
```

```python
new_student = pd.DataFrame([{
    "study_hours": 5,
    "attendance": 85,
    "previous_score": 75,
    "assignments_completed": 8
}])

predicted_marks = best_model.predict(new_student)[0]

predicted_marks = np.clip(predicted_marks, 0, 100)

print("Predicted exam marks:", round(predicted_marks, 2), "/ 100")
```

```python
import gradio as gr

def predict_student_score(
    study_hours,
    attendance,
    previous_score,
    assignments_completed
):
    input_data = pd.DataFrame([{
        "study_hours": study_hours,
        "attendance": attendance,
        "previous_score": previous_score,
        "assignments_completed": assignments_completed
    }])

    prediction = float(best_model.predict(input_data)[0])
    prediction = float(np.clip(prediction, 0, 100))

    return f"Predicted Exam Score: {prediction:.2f} / 100"

demo = gr.Interface(
    fn=predict_student_score,
    inputs=[
        gr.Slider(0, 10, value=5, step=0.5,
                  label="Study Hours Per Day"),
        gr.Slider(50, 100, value=85, step=1,
                  label="Attendance Percentage"),
        gr.Slider(0, 100, value=75, step=1,
                  label="Previous Exam Score"),
        gr.Slider(0, 10, value=8, step=1,
                  label="Assignments Completed")
    ],
    outputs=gr.Textbox(label="Prediction Result"),
    title="Student Performance Prediction System",
    description=(
        "Enter student details to estimate an exam score "
        "using a trained machine learning model."
    )
)

demo.launch(share=True)
```
